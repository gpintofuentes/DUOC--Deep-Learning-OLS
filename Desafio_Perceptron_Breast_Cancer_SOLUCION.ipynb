# SOLUCIÓN: Perceptrón — Dataset Breast Cancer
Versión resuelta del desafío. Clasifica tumores como **malignos (0)** o **benignos (1)** usando dos características: `mean radius` y `mean texture`.

## 1. Librerías

```python
import numpy as np
import matplotlib.pyplot as plt
import random
from sklearn import datasets
```

## 2. Función de activación y clase Perceptrón

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

class Perceptron():

    def __init__(self, input_size=2, lr=0.01, epochs=20):
        self.lr = lr
        self.epochs = epochs
        self.input_size = input_size
        self.w = np.random.uniform(-1, 1, size=(input_size))
        self.bias = random.uniform(-1, 1)
        self.misses = []

    def predict(self, X):
        z = sigmoid(np.dot(X, self.w) + self.bias)
        if z > 0.5:
            return 1
        else:
            return 0

    def fit(self, X, y):
        for epoch in range(self.epochs):
            miss = 0
            for yi, xi in zip(y, X):
                y_pred = self.predict(xi)
                error = yi - y_pred
                self.w += self.lr * error * xi
                self.bias += self.lr * error
                miss += int(error != 0.0)
            self.misses.append(miss)
```

## 3. Cargar el dataset

```python
data = datasets.load_breast_cancer()
X = data.data
y = data.target

print(data.feature_names[:5])
print(data.target_names)
print(X.shape, y.shape)
```

## 4. Elegir 2 características
Usamos `mean radius` (índice 0) y `mean texture` (índice 1).

```python
X = X[:, [0, 1]]
```

## 5. Graficar los datos

```python
plt.figure(figsize=(16, 8))
plt.scatter(X[y==0, 0], X[y==0, 1], marker='o', label='malignant')
plt.scatter(X[y==1, 0], X[y==1, 1], marker='x', label='benign')
plt.xlabel('mean radius')
plt.ylabel('mean texture')
plt.legend()
plt.show()
```

## 6. Entrenar el perceptrón

```python
perceptron = Perceptron()
perceptron.fit(X, y)
print(perceptron.w, perceptron.bias)
```

## 7. Graficar los errores por época

```python
plt.figure(figsize=(8, 6))
plt.plot(range(1, len(perceptron.misses) + 1), perceptron.misses, marker='o', linestyle='-')
plt.title('Errores de clasificación por época')
plt.xlabel('Época')
plt.ylabel('Número de errores de clasificación')
plt.grid()
plt.show()
```

**Nota:** a diferencia de IRIS (donde las clases setosa/versicolor son linealmente separables y el perceptrón llega a 0 errores), con `mean radius` y `mean texture` de Breast Cancer las clases se superponen bastante, así que es esperable que el perceptrón **no** converja a cero errores y la curva se estabilice en un valor mayor a 0. Esto es un buen disparador para hablar en clase sobre separabilidad lineal y las limitaciones del perceptrón simple.

## 8. Preguntas del desafío — comentarios de referencia
1. **Variar `lr`:** con `lr` muy chico (ej. 0.001) el perceptrón converge lento y puede no estabilizarse en 20 épocas; con `lr` grande (ej. 0.5) los pesos oscilan más y la curva de errores es más ruidosa. No hay un 'mejor' valor único, depende del trade-off velocidad/estabilidad.
2. **Cambiar features:** probando por ejemplo `mean perimeter` (índice 2) y `mean smoothness` (índice 4), la separación entre clases cambia — algunas combinaciones separan mejor que otras porque no todas las características son igual de discriminantes entre tumores malignos y benignos.
3. **Más `epochs`:** en general el número de errores baja rápido al principio y luego se estabiliza (plateau); agregar más épocas después de ese punto no mejora mucho porque el problema no es perfectamente separable con estas 2 features.

```python
# Ejemplo para la pregunta 1: comparar distintos learning rates
plt.figure(figsize=(8, 6))
for lr in [0.001, 0.01, 0.1, 0.5]:
    p = Perceptron(lr=lr)
    p.fit(X, y)
    plt.plot(range(1, len(p.misses) + 1), p.misses, marker='o', label=f'lr={lr}')
plt.xlabel('Época')
plt.ylabel('Errores de clasificación')
plt.title('Comparación de learning rates')
plt.legend()
plt.grid()
plt.show()
```
